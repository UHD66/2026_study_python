```python
from os import remove
import pandas as pd
import random as rd

statas={
    "NAME":["PLAYER","CPU"],
    "HP":[40,40],
    "MP":[10,10],
    "MONEY":[20,20]
}

ATK_card_select={
    "NAME":["a","b","c","d","e","f","g","h","i"],
    "DMG":[10,20,30,40,50,60,70,80,90],
    "MONEY":[3,4,5,6,7,8,9,10,11],
    "TYPE":[0,0,0,1,2,3,4,5,6]
}
#0=無属性、1=水,2=炎,3=土,4=木,5=光,6=闇
DEF_card_select={
    "NAME":["aa","bb","cc","dd","ee","ff","gg","hh","ii"],
    "DFD":[11,22,33,44,55,66,77,88,99],
    "MONEY":[1,2,3,4,5,6,7,8,9],
    "TYPE":[0,0,0,1,2,3,4,5,6]
}

other_card_select={
    "NAME":["ab","bc","cd","de","ef","fg","gh","hi","ij"]

}
statas_df=pd.DataFrame(statas)
ATK_card_df=pd.DataFrame(ATK_card_select)
DEF_card_df=pd.DataFrame(DEF_card_select)
other_card_df=pd.DataFrame(other_card_select)

#print(statas_df)
def draw_card():
  global ATK_player_card,DEF_player_card,ATK_card_select,DEF_card_select
  a=rd.randint(1,2)

  if a==1:
    card=rd.choice(ATK_card_select['NAME'])
    ATK_player_card.append(card)
    print(f"draw:{ATK_player_card}")
   # return card
  elif a==2:
    DEF_drop=rd.choice(DEF_card_select['NAME'])
    DEF_player_card.append(DEF_drop)
    print(f"draw{DEF_player_card}")
    #return DEF_drop
def cpu_draw():
  bb=rd.randint(1,2)
  if bb==1:
    b=rd.choice(DEF_card_select['NAME'])
    DEF_CPU_card.append(b)
  elif bb==2:
    b=rd.choice(ATK_card_select['NAME'])
    ATK_CPU_card.append(b)

ATK_card=rd.randint(1,9)
a=9-ATK_card
DEF_card=rd.randint(0,a)
other_card=9-ATK_card-DEF_card

ATK_card_df=ATK_card_df.sample(frac=1).reset_index(drop=True)
DEF_card_df=DEF_card_df.sample(frac=1).reset_index(drop=True)

#print(ATK_card_df)

ATK_player_card=ATK_card_df["NAME"][0:ATK_card].tolist()
#print(ATK_player_card)
DEF_player_card=DEF_card_df["NAME"][0:DEF_card].tolist()
#print(DEF_player_card)
player_other_card=other_card_df["NAME"][0:other_card].tolist()
#print("="*30)

CPU_ATK_card=rd.randint(1,9)
b=9-CPU_ATK_card
CPU_DEF_card=rd.randint(0,b)
CPU_other_card=9-CPU_ATK_card-CPU_DEF_card

ATK_CPU_card=ATK_card_df["NAME"][0:CPU_ATK_card].tolist()
#print(ATK_CPU_card)
DEF_CPU_card=DEF_card_df["NAME"][0:CPU_DEF_card].tolist()
print(DEF_CPU_card)
cpu_other_card=other_card_df["NAME"][0:CPU_other_card].tolist()
#print(cpu_other_card)
#print("="*30)

while True:
#手札表示
  print("="*10,"ステータス","="*10)
  print(statas_df)
  print("="*10,"手札","="*10)
  print(f"番号、名前、攻撃力、価格")
# enumerateは、cardinfoの中身に数字を割り当ててくれるやつ
  for i,card_name in enumerate(ATK_player_card):
    card_info=ATK_card_df[ATK_card_df['NAME']==card_name]

    print(f"[{i}] {card_info.values}")
  print("="*30)

  for i,def_card_name in enumerate(DEF_player_card):
    def_card_info=DEF_card_df[DEF_card_df["NAME"]==def_card_name]

    print(f"[{i}] {def_card_info.values}")

  print("="*30)
  player_turn=int(input())

  if len(DEF_CPU_card)>0:
    a=rd.choice(DEF_CPU_card)
    print(f"CPUのだした防御カード:{a}")
    card_index = DEF_card_select["NAME"].index(a)
    damage=max(0,ATK_card_df.loc[player_turn,"DMG"]-DEF_card_df.loc[card_index,"DFD"])
    DEF_CPU_card.remove(a)
    cpu_draw()

    if damage>0 and len(DEF_CPU_card)>0:
      a=rd.choice(DEF_CPU_card)
      print(f"CPUのだした防御カード:{a}")
      card_index = DEF_card_select["NAME"].index(a)
      damage=max(0,ATK_card_df.loc[player_turn,"DMG"]-DEF_card_df.loc[card_index,"DFD"])
      statas_df.loc[1,"HP"]=statas_df.loc[1,"HP"]-damage
      DEF_CPU_card.remove(a)
      cpu_draw()

    player_card_remove=ATK_player_card[player_turn]
    #つかったカード消去
    ATK_player_card.remove(player_card_remove)


    bb=rd.randint(1,2)
    if bb==1:
      b=rd.choice(DEF_card_select['NAME'])
      DEF_CPU_card.append(b)
    elif bb==2:
      b=rd.choice(ATK_card_select['NAME'])
      ATK_CPU_card.append(b)
    print(f"CPUの防御:{DEF_card_df.loc[card_index,"DFD"]}")
    print(f"CPUHP:{statas_df.loc[1,"HP"]}")
    #break

  else:
    statas_df.loc[1,"HP"]=statas_df.loc[1,"HP"]-ATK_card_df.loc[player_turn,"DMG"]
    print(f"CPUHP:{statas_df.loc[1,"HP"]}")

    player_card_remove=ATK_player_card[player_turn]
    ATK_player_card.remove(player_card_remove)

    #break

#CPUの攻撃
  print("="*40)
  if len(DEF_player_card)>0:
    for i,def_card_name in enumerate(DEF_player_card):
      def_card_info=DEF_card_df[DEF_card_df["NAME"]==def_card_name]

      print(f"[{i}] {def_card_info.values}")

    CPU_attack=rd.choice(ATK_CPU_card)


    damage=ATK_card_df.loc[ATK_card_df["NAME"]==CPU_attack,"DMG"].values[0]
    print(f"出したカード{CPU_attack},攻撃力:{damage}")
    player_def=int(input())

    damage=max(0,damage-DEF_card_df.loc[player_def,"DFD"])
    print(damage)
    #statas_df.loc[0,"HP"]=statas_df.loc[0,"HP"]-damage

    remove_card=DEF_player_card[player_def]
    DEF_player_card.remove(remove_card)
    print(DEF_player_card)

    draw_card()


    if damage>0 and len(DEF_player_card)>0:
      for i,def_card_name in enumerate(DEF_player_card):
        def_card_info=DEF_card_df[DEF_card_df["NAME"]==def_card_name]
        print(f"[{i}] {def_card_info.values}")

      player_def=int(input())
      damage=max(0,damage-DEF_card_df.loc[player_def,"DFD"])
      print("="*40)
      print(f"CPUから{damage}ダメージ")

      remove_card=DEF_player_card[player_def]
      DEF_player_card.remove(remove_card)
      print(DEF_player_card)

      draw_card()
    elif damage<=0 and len(DEF_player_card)<=0:
      print("="*40)
      print(f"CPUから{damage}ダメージ")

    statas_df.loc[0,"HP"]=statas_df.loc[0,"HP"]-damage

    print("="*40)
    print(f"CPUから{damage}ダメージ")
    print("="*40)
    #print(statas_df)
    #break
  else:
    CPU_attack=rd.choice(ATK_CPU_card)

    damage=ATK_card_df.loc[ATK_card_df["NAME"]==CPU_attack,"DMG"].values[0]
    print(f"出したカード{CPU_attack},攻撃力:{damage}")
    statas_df.loc[0,"HP"]=statas_df.loc[0,"HP"]-damage
    print(f"CPUから{damage}ダメージ")
```
