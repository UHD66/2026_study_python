```python
import random

class Omikuzi:
  def __init__(self, name):
    self.name=name
    self.result=["大吉", "中吉", "小吉", "凶"]

  def draw(self):
    a=random.choice(self.result)
    print(f"{self.name}さんの運勢は{a}です")

user1=Omikuzi("oda")
user1.draw()

user2=Omikuzi("oda2")
user2.draw()
```
