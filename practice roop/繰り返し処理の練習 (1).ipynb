```python
import random
count=0
cpu=0
A=["グー","チョキ","パー"]
a=random.choice(A)

for i in range(3):
  b=input()
  if a==b:
    print("あいこ")
  elif a=="グー" and b=="チョキ" or a=="パー"and b=="グー" or a=="チョキ"and b=="パー":
    print("負け")
    cpu=cpu+1
  else:
    print("勝ち")
    count=count+1
  print(f"cpuの手は{a}")
print(f"あなたの勝利回数は{count}で,cpuの勝利回数は{cpu}でした！")
```
